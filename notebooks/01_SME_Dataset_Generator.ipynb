```python
def generate_employee():

    department = random.choice([
        "HR", "Finance", "IT", "Marketing", "Legal"
    ])

    employee = {
        "EmployeeID": fake.random_int(1000, 9999),
        "Department": department,
        "DocumentType": "Employee Record"
    }

    # Generate different data profiles
    profile = random.choice([
        "Public",
        "Internal",
        "Confidential",
        "Restricted"
    ])

    # Public: No sensitive personal or financial information
    if profile == "Public":
        pass

    # Internal: Internal business information only
    elif profile == "Internal":
        employee["InternalNote"] = "Department operational record"

    # Confidential: Personal or sensitive information
    elif profile == "Confidential":
        employee.update({
            "Name": fake.name(),
            "NIC": str(fake.random_number(digits=12)),
            "Email": fake.email(),
            "Salary": random.randint(50000, 300000)
        })

    # Restricted: Combination of highly sensitive information
    elif profile == "Restricted":
        employee.update({
            "Name": fake.name(),
            "NIC": str(fake.random_number(digits=12)),
            "Email": fake.email(),
            "Salary": random.randint(50000, 300000),
            "BankAccount": str(fake.random_number(digits=10))
        })

    # Determine classification using actual field combinations
    has_nic = "NIC" in employee
    has_email = "Email" in employee
    has_salary = "Salary" in employee
    has_bank = "BankAccount" in employee

    if has_nic and has_bank and has_salary:
        classification = "Restricted"

    elif has_nic or has_email or has_salary:
        classification = "Confidential"

    elif "InternalNote" in employee:
        classification = "Internal"

    else:
        classification = "Public"

    # Assign risk level and leakage score
    risk_mapping = {
        "Public": ("Low", (1, 25)),
        "Internal": ("Medium", (26, 50)),
        "Confidential": ("High", (51, 80)),
        "Restricted": ("Critical", (81, 100))
    }

    risk_level, score_range = risk_mapping[classification]

    employee["Classification"] = classification
    employee["RiskLevel"] = risk_level
    employee["LeakageScore"] = random.randint(*score_range)

    return employee


print("Rule-Based Employee Dataset Generator Ready!")
```

```python
employees = []

for i in range(500):
    employees.append(generate_employee())

df = pd.DataFrame(employees)

print("Total Records:", len(df))
display(df.head())
```

```python
risk_mapping = {
    "Public": "Low",
    "Internal": "Medium",
    "Confidential": "High",
    "Restricted": "Critical"
}

df["ExpectedRisk"] = df["Classification"].map(risk_mapping)

mismatches = df[
    df["RiskLevel"] != df["ExpectedRisk"]
]

print("Risk mismatches:", len(mismatches))
```
